# Static image

This is an example of applying retinomorphic convolution to a static image. It also demonstrates how the different layers are built and shows some plots of cells and their receptive fields for each layer.

```python
# --------------------------------------
import warnings

warnings.filterwarnings("ignore")

# --------------------------------------
from pathlib import Path

# --------------------------------------
import numpy as np

# --------------------------------------
import skimage as ski

# # --------------------------------------
import pyrception as pcp
from pyrception.visual.layers import ReceptorLayer
from pyrception.visual.layers import HorizontalLayer
from pyrception.visual.layers import BipolarLayer
from pyrception.visual.layers import AmacrineLayer
from pyrception.visual.layers import GanglionLayer
from pyrception.utils.plotting import Plotter
```

## Open an image

```python
img_path = "https://upload.wikimedia.org/wikipedia/commons/b/b6/Felis_catus-cat_on_snow.jpg"
img = ski.io.imread(img_path, as_gray=True)
img = ski.transform.resize(img, np.array(img.shape) // 5, anti_aliasing=True)
```

```python
size = img.shape
h, w = size
h, w
```

## Visualising the output all retinal layers

```python
pl = Plotter()
```

Show the image that we are working with.

```python
p = pl.image(img, title="Original image (greyscale)")
pl.show(p)
```

Convenience functions for plotting parameters

## Receptor layer

```python
rl_params = {
    "shape": img.shape,
}

rl = ReceptorLayer(**rl_params)
```

```python
rl_activation = rl.forward(img)
```

## Horizontal layer

```python
hl_params = {
    "shape": img.shape,
    "receptor": rl,
    "sectors": 64,
    "rf_params": {
        "shape": pcp.KernelShape.Elliptic,
        "filter": pcp.KernelFilter.Uniform,
        "min_size": (3,3),
        "scale": (1.0, 1.0),
        "angle": 0,
    },
}

if hl_params['rf_params']["filter"] == pcp.KernelFilter.Gaussian:
    hl_params['rf_params']["kernel_params"] = {"normalise": True}

hl = HorizontalLayer(**hl_params)
```

```python
rbegin = 0
rend = 50
hl_ring_canvas = hl.plot_rfs(cells=hl.rfs.cell_rings[rbegin:rend])
hl_ring_plot = pl.image(hl_ring_canvas)
pl.show(
    hl_ring_plot, title=f"Horizontal layer receptive fields (rings {rbegin}-{rend})"
)
```

```python
sbegin = 0
send = 10
hl_sector_canvas = hl.plot_rfs(cells=hl.rfs.cell_sectors[sbegin:send])
hl_sector_plot = pl.image(hl_sector_canvas)
pl.show(
    hl_sector_plot, title=f"Horizontal layer receptive fields (sectors {sbegin}-{send})"
)
```

```python
(hl_activation, hl_feedback) = hl.forward()
hl_feedback = hl_feedback.reshape(hl.rfs.height, hl.rfs.width)
hl_feedback_canvas = pl.image(hl_feedback)
pl.show(hl_feedback_canvas)
```

Process the image with the horizontal layer and plot the activations as a scatter plot.

```python
scat = pl.scatter(hl_activation, xtitle="Neuron", ytitle="Activation", width=1000)

pl.show(scat)
```

```python
print(f"==[ receptor activation: {rl.membrane} | {rl.membrane.min(), rl.membrane.max()}")
print(f"==[ horizontal activation: {hl.activation} | {hl.activation.min(), hl.activation.max()}")
```

```python
hl_act_canvas = np.zeros((h, w))

hl_act_canvas[
    hl.rfs.cell_coordinates[:, 0],
    hl.rfs.cell_coordinates[:, 1],
] = hl_activation

hl_act_canvas = pl.image(hl_act_canvas)
pl.show(hl_act_canvas)
```

Compute the normalised version of the raw input by subtracting the mean illumination map. The normalised input is then passed to the bipolar layer.

```python
normalised_canvas = img - hl_feedback.reshape((h, w))
normalised_canvas = ski.exposure.rescale_intensity(normalised_canvas, out_range=(0.0,1.0))

scaled_plot = pl.image(normalised_canvas, title="Normalised input")

pl.show(scaled_plot)
```

## Bipolar layer

```python
bl_params = {
    "shape": size,
    "receptor": rl,
    "horizontal": hl,
    "sectors": 96,
    "rf_params": {
        "shape": pcp.KernelShape.Elliptic,
        "filter": pcp.KernelFilter.Uniform,
        "scale": 1.0,
    },
}

bl = BipolarLayer(**bl_params)
```

```python
rbegin = 30
rend = 50

bl_ring_canvas = bl.plot_rfs(cells=bl.rfs.cell_rings[rbegin:rend])
bl_ring_plot = pl.image(bl_ring_canvas)
pl.show(
    bl_ring_plot,
    title=f"Bipolar layer receptive fields (rings {rbegin}-{rend})",
)
```

```python
sbegin = 0
send = 10
bl_sector_canvas = bl.plot_rfs(cells=bl.rfs.cell_sectors[sbegin:send])
bl_sector_plot = pl.image(bl_sector_canvas)
pl.show(
    bl_sector_plot, title=f"Bipolar layer receptive fields (sectors {sbegin}-{send})"
)
```

```python
bl_out = bl.forward()
```

```python
bl_out.min(), bl_out.max()
```

```python
bp_on_canvas = np.zeros((h, w))
bp_off_canvas = np.zeros((h, w))

bp_on_canvas[bl.rfs.cell_coordinates[:, 0], bl.rfs.cell_coordinates[:, 1]] = bl_out

on_plot = pl.image(bp_on_canvas, title="ON bipolar")

pl.show([[on_plot]])
```

## Amacrine layer

```python
al_params = {
    "shape": size,
    "bipolar": bl,
    "sectors": 112,
    "rf_params": {
        "filter": pcp.KernelFilter.Uniform,
        "min_size": 3,
        "scale": 1.5,
        "phyllotactic": False,
    },
}

al = AmacrineLayer(**al_params)
```

```python
rbegin = 0
rend = None

al_canvas = al.plot_rfs(
    cells=[al.rfs.cell_rings[rbegin:rend]],
    cell_colour="#00ffff",
    rf_colour="#ff00ff",
)
```

```python
rbegin = 0
rend = 50

al_ring_canvas = al.plot_rfs(cells=al.rfs.cell_rings[rbegin:rend])
al_ring_plot = pl.image(al_ring_canvas)
pl.show(
    al_ring_plot,
    title=f"Amacrine layer receptive fields (rings {rbegin}-{rend})",
)
```

Receptive field for a single amacrine cell (highlights the sparsity of the input)

```python
title = (
    f"Amacrine layer receptive fields (rings {rbegin}-{rend} and sectors {sbegin}-{send})",
)
```

```python
sbegin = 0
send = 10
al_sector_canvas = al.plot_rfs(cells=al.rfs.cell_sectors[sbegin:send])
al_sector_plot = pl.image(al_sector_canvas)
pl.show(
    al_sector_plot, title=f"Amacrine layer receptive fields (sectors {sbegin}-{send})"
)
```

```python
cell = 400
al_single_canvas = al.plot_rfs(cells=cell)
al_single_plot = pl.image(al_single_canvas)
pl.show(al_single_plot, title=f"Amacrine layer receptive fields (cell {cell})")
```

```python
al_on = al.forward()
```

```python
al_on_canvas = np.zeros((h, w))
al_off_canvas = np.zeros((h, w))

al_on_canvas[al.rfs.cell_coordinates[:, 0], al.rfs.cell_coordinates[:, 1]] = al_on

al_on_plot = pl.image(al_on_canvas, title="ON amacrine")

pl.show([[al_on_plot]])
```

## Ganglion layer

```python
gl_bl_params = {
    "scale": 2.0,
    "filter": pcp.KernelFilter.Uniform,
}
gl_al_params = {
    "scale": 5.0,
    "filter": pcp.KernelFilter.Uniform,
}

gl_params = {
    "shape": size,
    "bipolar": bl,
    "amacrine": al,
    "sectors": 128,
    "bipolar_params": gl_bl_params,
    "amacrine_params": gl_al_params,
    "threshold": 0.01,
}

gl = GanglionLayer(**gl_params)
```

```python
rbegin = 0
rend = None

gl_ring_canvas = gl.plot_rfs(cells=gl.amacrine_rfs.cell_rings[rbegin:rend])
gl_ring_plot = pl.image(gl_ring_canvas)
pl.show(
    gl_ring_plot,
    title=f"Amacrine layer receptive fields (rings {rbegin}-{rend})",
)
```

```python
gl.membrane.min(), gl.membrane.max()
```

```python
gl_on_off = gl.forward()
```

```python
gl_on_off_canvas = np.zeros((h, w))

gl_on_off_canvas[
    gl.amacrine_rfs.cell_coordinates[:, 0],
    gl.amacrine_rfs.cell_coordinates[:, 1],
] = gl_on_off

gl_on_off_plot = pl.image(gl_on_off_canvas, title="ON-centre/OFF-surround ganglion")

pl.show([[gl_on_off_plot]])
```

```python
print(f"==[ Active ganglion cells: {gl_on_off.sum()} / {len(gl_on_off)}")
```
